### Read sample1.txt file into RDD and displaying the first 4 elements

```python
text = '''Utilitatis causa amicitia est quaesita.
Lorem ipsum dolor sit amet, consectetur adipiscing elit.
Collatio igitur ista tenihil iuvat.
Honesta oratio, Socratica, Platonis etiam.
Primum in nostranepotestate est, quid meminerimus?
Duo Reges: constructio interrete.
Quid, sietiam iucunda memoria est praeteritorum malorum?
Si quidem, inquit, tollerem,'''

with open('/content/sample1.txt', 'w') as f:
    f.write(text)
```

```python
file_read = sc.textFile("/content/sample1.txt")
```

```python
file_read.take(4)
```

### Count the total number of rows in RDD

```python
total_rows = file_read.count()
print(total_rows)
```

### Create a function to convert the data into lower case and splitting it

```python
def lower_split(text):
  return text.lower().split()

word_rdd = file_read.flatMap(lower_split)
word_rdd.collect()
```

```python
lines_lowered = file_read.map(lower_split)
lines_lowered.collect()
```

### Remove the stopwords from the previous text. i.e. Remove it.

```python
stopwords = ['a','all','the','as','is','am','an','and',
             'be','been','from','had','I','I’d','why','with']
# Hint: you may need use flatMap
```

```python
filtered_words = word_rdd.filter(lambda word: word not in stopwords)
filtered_words.take(5)
```

```python
# using flatmap will be on original text
filt_fm_rdd = file_read.flatMap(lambda line: line.split()).filter(lambda word: word not in stopwords)
filt_fm_rdd.take(5)
```

### Find the words starting with ‘c’

```python
c_words_rdd = filtered_words.filter(lambda word: word.startswith('c'))
c_words_rdd.collect()
```

### Reduce the data by key and sum it (use the data from the following list)

```python
list = [('JK', 22), ('V', 24), ('Jimin',24), ('RM', 25)
        , ('J-Hope', 25), ('Suga', 26), ('Jin', 27)
       , ('J-Hope', 12), ('Suga', 25), ('Jin', 34)
       , ('JK', 32), ('V', 44), ('Jimin',14), ('RM', 35)]
# Hint: use reduceByKey
```

```python
rdd2 = sc.parallelize(list)
```

```python
reduced_data = rdd2.reduceByKey(lambda x, y: x + y)
reduced_data.collect()
```

### Creat some key value pairs RDDs

```python
rdd4 = sc.parallelize([('a',2),('b',3)])
rdd5 = sc.parallelize([('a',9),('b',7),('c',10)])
```

### Perform Join operation on the RDDs (rdd1,rdd2)

```python
join_rdd = rdd4.join(rdd5)
join_rdd.collect()
```
