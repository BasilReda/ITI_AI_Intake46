```python
rddfile_split_flat = rddfile.flatMap(lambda x : x.split(','))
```

```python
rddfile_split_flat_2 = rddfile_split_flat.flatMap(lambda x: x.split(" ")).filter(lambda x: x!= " ").filter(lambda x: x!="")
```

```python
rddfile_split_flat_2.take(15)
```

```python
rddfile_split_flat_3 =  rddfile_split_flat_2.keyBy(lambda x : x[0])
```

```python
rddfile_split_flat_3.take(1000)
```

```python
rdd_groups = rddfile_split_flat_3.groupByKey()
```

```python
rdd_groups.take(11)
```

```python
rdd_readable2 = rdd_groups.map(lambda x : (x[0], list(x[1])))

rdd_readable2.take(1)
```

```python
rdd_readable = rdd_groups.mapValues(list)

rdd_readable.take(1)
```

```python
dataRDD =sc.parallelize( [
    ("Ali", 25),
    ("Mona", 30),
    ("Esraa", 28),
    ("Ali", 27),
    ("Mona", 32),
    ("Mariam", 24),
    ("Esraa", 29),
    ("Ali", 26)
])
```

```python
dataRDD2 = dataRDD.groupByKey().mapValues(lambda x: sum(list(x))/len(list(x)))
dataRDD2.collect()
```

```python
dataRDD3 = dataRDD.groupByKey().map(lambda x: (x[0], sum(list(x[1]))/len(list(x[1]))))
dataRDD3.collect()
```

```python
dataRDD4 = dataRDD.reduceByKey(lambda x, y: x + y)
dataRDD4.collect()
```
