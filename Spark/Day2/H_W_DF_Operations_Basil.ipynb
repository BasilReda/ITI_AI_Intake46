$$Bonus Question!$$
> - There are `too many decimal places` for mean and stddev in the describe() dataframe. <br>
- Format the numbers to just show up to `two decimal places.`
- Pay careful attention to the datatypes that .describe() returns, we didn't cover how to do this exact formatting, but we covered something very similar.
- If you get stuck on this, don't worry, just view the solutions.

```python
from pyspark.sql import functions as F
summary_df = df.describe()
formatted_summary = summary_df.select(
    "summary",
    *[F.round(F.col(c).cast("double"), 2).alias(c) for c in summary_df.columns if c != "summary"]
)
formatted_summary.show()
```

`###` Create a new dataframe with a column called `HV Ratio` that is the ratio of the `High Price versus volume of stock traded for a day`.

```python
hv_ratio = df.withColumn("HV Ratio", df["High"] / df["Volume"])
hv_ratio.select("HV Ratio").show()
```

`#` **What day had the `Peak High` in `Price`?**

```python
peak_high = df.orderBy(df["High"].desc()).first()
peak_high_date = peak_high["Date"]
print(peak_high_date)
```

`##` **What is the `mean` of the `Close` column?**

```python
df.select(mean("Close")).show()
```

`###` **What is the `max` and `min` of the `Volume` column?**

```python
df.select(F.min("Volume"), F.max("Volume")).show()
```

`####` **How many days was the Close `lower than 60` dollars?**

```python
df.filter(df["Close"] < 60).count()
```

```python
df.select(F.sum((df["Close"] < 60).cast("int"))).show()
```

`#####` **What `percentage of the time` was the `High greater than 80`dollars ?**
-  In other words, (Number of Days High>80)/(Total Days in the dataset)

```python
(df.filter(df["High"] > 80).count() / df.count()) * 100
```

`#####` **What is the `Pearson correlation` between `High and Volume?`**

```python
df.select(F.corr("High", "Volume")).show()
```

`######` **What is the `max High` per year?**

```python
df.withColumn("Year", F.year("Date")).groupBy("Year").max("High").show()
```

`####` What is the `average Close` for each `Calendar Month`?
  - In other words, across all the years, what is the average Close price for `Jan,Feb, Mar, etc...` Your result will have a value for each of these months.

```python
df.withColumn("Month", F.month("Date")).groupBy("Month").avg("Close").orderBy("Month").show()
```

`i` Read `"test1"` dataset

```python
filepath = "/content/drive/MyDrive/test1.csv"
df1 = spark.read.csv(filepath, header=True, inferSchema=True)
df1.show()
```

`ii` Display `Salary` of the people `less than or equal to 20000`

```python
from pyspark.sql import functions as F
df1.withColumn(
    "Salary",
    F.when(F.col("Salary") <= 20000, F.col("Salary"))
).select("Salary").dropna().show()
```

`iii` Display `Salary` of the people `less than or equal to 20000` and `greater than or equal 15000`

```python
df1.withColumn(
    "Salary",
    F.when((F.col("Salary") <= 20000) & (F.col("Salary") >= 15000), F.col("Salary"))
).select("Salary").dropna().show()
```
