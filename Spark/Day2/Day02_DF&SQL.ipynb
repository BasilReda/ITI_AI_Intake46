## Dealing with Null data

```python
dfNull = spark.read.csv('/content/drive/My Drive/pyspark/Data/NullData.csv',inferSchema=True,header=True)
```

```python
dfNull.show()
```

```python
df_null = dfNull.withColumn('DoubleSale',expr(' Sales * 2 as dblsale'))
```

```python
df_null.show()
```

```python
df_notnull = df_null.na.drop()
```

```python
df_notnull.show()
```

```python
df_notnull2 = df_null.dropna()
```

```python
df_notnull2.show()
```

```python
df_notnull = df_null.na.drop(how='all')
```

```python
df_notnull.show()
```

```python
df_notnull = df_null.dropna(subset=['Name','Sales','DoubleSale'],how='all')
df_notnull.show()
```

```python
df_null.show()
```

```python
df_notnull = df_null.na.drop(thresh=2)
df_notnull.show()
```

```python
df_notnull = df_null.na.drop(thresh=3)
df_notnull.show()
```

```python
df_null.show()
```

```python
df_notnull = df_null.na.drop(subset=['Name','Sales','DoubleSale'],thresh=2)
df_notnull.show()
```
