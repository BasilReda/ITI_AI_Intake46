### ITI Day 1 Lab Challenge: Real-Time Pipeline Scaling & Replay

**Scenario:** You are a Data Engineer building a real-time ingestion pipeline for a bank. Transactions are flooding into the cluster, and you need to deploy a scalable, fault-tolerant consumer group to process both historical logs and live data streams in parallel.

## Your Tasks:

### **Task 1: The Producer (Data Source)**

- Open a dedicated terminal tab.
- Start a console producer targeting the `raw-transactions` topic.
- Type 6 distinct historical messages rapidly into the prompt (e.g., `tx_1`, `tx_2`, `tx_3`, `tx_4`, `tx_5`, `tx_6`) and leave the producer open.

### **Task 2: Parallel Consumer Group Setup**

- Open **two separate terminal windows side-by-side**.
- Keep them clear and ready; do not press enter on any commands yet.

### **Task 3: Run the Load-Balancing Test (From the Beginning)**

- In **both** consumer windows, type the console consumer command targeting the `raw-transactions` topic.
- **Crucial Additions:** Ensure both consumers belong to the same group named `fraud-analysis-team` **AND** include the `--from-beginning` flag.
- Run the commands at the same time and observe your screens.
- **Observation:** Notice how Kafka dynamically reads all 6 historical messages from disk and splits them evenly between your two active consumer windows.

### **Task 4: Simulate a Production Crash & Recovery**

- Forcefully shut down **one** of your active consumers (`Ctrl + C`).
- Go back to your Producer terminal (Task 1) and type 3 new live messages (`tx_7`, `tx_8`, `tx_9`).
- **Observation:** Verify that the single surviving consumer automatically detects the failure and instantly takes over the remaining workload without missing a single transaction.
