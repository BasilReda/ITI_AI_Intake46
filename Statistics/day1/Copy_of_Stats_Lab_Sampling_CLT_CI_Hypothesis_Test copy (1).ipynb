# 🪙Sampling Distributions and CLT
- Take a `10000` samples of sizes `1, 5, 10, 50` from a continuous `uniform distribution` `a=1, b=11`. **Plot the sampling distribution of the sample mean.**
- Calculate `mean` and `variance` of sampling distribution.
- Plot a histogram for the sampling distribution.
- Repeat for any distribution of your choice.

Hint: fix the scale of x-axis to be able to notice the difference.

```python
import numpy as np
from scipy import stats as sp_stats

# Sampling Distribution with Uniform Distribution (a=1, b=11)
num_samples = 10000
a, b = 1, 11
pop_mean = (a + b) / 2
pop_variance = ((b - a) ** 2) / 12

# Sample size n=1
sample_size = 1
sample_means_1 = [np.mean(np.random.uniform(a, b, sample_size)) for _ in range(num_samples)]
mean_1, var_1 = np.mean(sample_means_1), np.var(sample_means_1)
print(f"n={sample_size:2d}  |  Mean: {mean_1:.4f}  |  Variance: {var_1:.4f}")
plt.figure(figsize=(10, 4))
plt.hist(sample_means_1, bins=50, alpha=0.7, edgecolor='black')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')
plt.title(f'Sampling Distribution (n={sample_size}, Uniform Distribution)')
plt.axvline(pop_mean, color='r', linestyle='--', label=f'Population Mean={pop_mean}')
plt.legend()
plt.tight_layout()
plt.show()
```

# 🪙 Confidence Interval (CI)
 Within how many standard deviations the estimated sample mean should be located for CIs `90%, 92%, 95%, 98%, 99%`.   

```python
from scipy.stats import norm, t as t_dist

# Calculate z-values for different confidence levels
confidence_levels = [0.90, 0.92, 0.95, 0.98, 0.99]
print("Confidence Level | Z-value (# of std devs)")
print("-" * 45)
for cl in confidence_levels:
    alpha = 1 - cl
    z_val = norm.ppf(1 - alpha/2)
    print(f"      {cl*100:.0f}%        |       {z_val:.4f}")
```

### Calculate the expected mean salary with `95% CI` for each team for the employees data as follow:
- Using z-value.
- Using t-value.
- Compare the results obtaind from the above calculation and write a conclusion.
- Perform  sns.barplot() and check out the displayed confidence intervals. Compare with your results.

```python
# Calculate 95% CI for each team using Z-value
salary_by_team = data_clean.groupby('Team')['Salary']
confidence_level = 0.95
alpha = 1 - confidence_level
z_val = norm.ppf(1 - alpha/2)

results = []
for team, group in salary_by_team:
    n = len(group)
    mean = group.mean()
    std = group.std()
    se = std / np.sqrt(n)
    ci_lower = mean - z_val * se
    ci_upper = mean + z_val * se
    t_val = t_dist.ppf(1 - alpha/2, df=n-1)
    t_ci_lower = mean - t_val * se
    t_ci_upper = mean + t_val * se
    results.append({
        'Team': team,
        'n': n,
        'Mean': mean,
        'Z-CI Lower': ci_lower,
        'Z-CI Upper': ci_upper,
        'T-CI Lower': t_ci_lower,
        'T-CI Upper': t_ci_upper
    })

import pandas as pd
results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))
```

```python
# Seaborn barplot with 95% CI
plt.figure(figsize=(12, 6))
sns.barplot(data=data_clean, x='Team', y='Salary', ci=95)
plt.title('Average Salary by Team with 95% Confidence Intervals')
plt.xlabel('Team')
plt.ylabel('Salary')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nComparison: Manual calculations match the seaborn barplot CI")
```
